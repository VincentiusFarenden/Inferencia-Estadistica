# **Introducción**

En estadística inferencial, el contraste de hipótesis permite tomar decisiones fundamentadas a partir de información muestral. En muchos contextos reales, el interés no se centra en describir una sola población, sino en **comparar dos poblaciones o dos grupos** con el fin de determinar si existen diferencias significativas entre ellas.

En este laboratorio se aborda el contraste de hipótesis en situaciones de comparación, considerando dos parámetros de interés ampliamente utilizados: la **diferencia de medias poblacionales** y la **diferencia de proporciones poblacionales**. Estos procedimientos permiten evaluar, por ejemplo, si dos grupos presentan rendimientos distintos, si un tratamiento produce efectos diferentes respecto de otro o si la proporción de ocurrencia de un evento varía entre dos poblaciones.

Si bien el tipo de variable y el estadístico de prueba pueden cambiar según el caso, ambos contrastes comparten una **misma estructura inferencial**: formulación de hipótesis nula y alternativa, selección del estadístico de prueba adecuado, cálculo del valor $p$ y toma de decisiones basada en un nivel de significancia previamente definido.

El objetivo de este laboratorio es que comprendas y apliques el contraste de hipótesis para la diferencia de medias y la diferencia de proporciones, interpretando correctamente los resultados obtenidos y relacionándolos con situaciones reales provenientes de ámbitos como la educación, la salud, la industria y las ciencias sociales.

## **Recordatorio clave**

### Hipótesis estadísticas en la comparación de dos poblaciones

En problemas de comparación entre dos poblaciones independientes, el contraste de hipótesis se centra en evaluar si existe una diferencia significativa entre los parámetros de interés.

- **Hipótesis nula $(H_{0})$**: no existe diferencia entre las poblaciones.
- **Hipótesis alternativa $(H_{1})$**: existe una diferencia entre las poblaciones, la cual puede ser **bilateral** o **unilateral**, según el contexto del problema.
<br>

### Contraste para la diferencia de medias poblacionales

**Hipótesis generales:**

$$H_{0}: \ \mu_1 = \mu_2 \quad \text{o} \quad H_{0}:\mu_1 \leq \mu_2 \quad \text{o} \quad H_{0}:\mu_1 \geq \mu_2$$

$$H_{1}: \ \mu_1 \neq \mu_2 \quad \text{o} \quad H_{1}:\mu_1 > \mu_2 \quad \text{o} \quad H_{1}:\mu_1 < \mu_2$$
<br>

#### Estadístico de prueba

En este laboratorio, para contrastar hipótesis sobre la diferencia entre dos medias de muestras independientes, se utilizará la **prueba $t$ de Welch**.

El estadístico de prueba está dado por:

$$t = \frac{\bar{X}_1 - \bar{X}_2}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}$$

con grados de libertad aproximados:

$$gl =
\frac{
\left( \frac{s_1^2}{n_1} + \frac{s_2^2}{n_2} \right)^2
}{
\frac{\left( \frac{s_1^2}{n_1} \right)^2}{n_1 - 1} +
\frac{\left( \frac{s_2^2}{n_2} \right)^2}{n_2 - 1}
}$$

**Nota:** En algunos contextos, cuando se puede asumir igualdad de varianzas poblacionales, puede utilizarse la prueba t de Student. Sin embargo, en este laboratorio trabajaremos únicamente con la prueba t de Welch.
<br>

### Contraste para la diferencia de proporciones poblacionales

**Hipótesis generales:**

$$H_{0}: \ p_1 = p_2 \quad \text{o} \quad H_{0}:p_1 \leq p_2 \quad \text{o} \quad H_{0}:p_1 \geq p_2$$

$$H_{1}: \ p_1 \neq p_2 \quad \text{o} \quad H_{1}:p_1 > p_2 \quad \text{o} \quad H_{1}:p_1 < p_2$$

#### Estadístico de prueba

Las proporciones muestrales se definen como:

$$\hat{p}_1 = \frac{x_1}{n_1},
\qquad
\hat{p}_2 = \frac{x_2}{n_2}$$

Bajo la hipótesis nula, se utiliza la **proporción combinada**:

$$\hat{p} = \frac{x_1 + x_2}{n_1 + n_2}$$

El estadístico de prueba es:

$$z = \frac{\hat{p}_1 - \hat{p}_2}
{\sqrt{\hat{p}(1 - \hat{p})\left(\frac{1}{n_1} + \frac{1}{n_2}\right)}}$$
<br>

### Regla de decisión

Para ambos contrastes:

- Si **valor $p$ $<$ $\alpha$** → se **rechaza $H_{0}$**.
- Si **valor $p$ $≥$ $\alpha$** → **no se rechaza $H_{0}$**.

donde $\alpha$ corresponde al **nivel de significancia** previamente establecido.
<br>

### Errores tipo I y tipo II

En el contraste de hipótesis para la comparación entre dos poblaciones, pueden cometerse dos tipos de errores en la toma de decisiones:

- **Error tipo I $(\alpha)$**: rechazar la hipótesis nula cuando en realidad **no existe diferencia** entre las poblaciones (por ejemplo, $\mu_1 = \mu_2$ o $p_1 = p_2$).
- **Error tipo II $(\beta)$**: no rechazar la hipótesis nula cuando en realidad **sí existe una diferencia** entre las poblaciones (por ejemplo, $\mu_1 \neq \mu_2$ o $p_1 \neq p_2$).

El valor de $\alpha$ corresponde al **nivel de significancia** fijado previamente y representa la probabilidad de cometer un error tipo I. El complemento $1 - \beta$ se denomina **poder de la prueba** y mide la capacidad del contraste para detectar una diferencia real cuando esta existe.

*Nota:* Aumentar el **tamaño muestral** contribuye a disminuir la probabilidad de cometer errores tipo I y tipo II, y a incrementar el poder de la prueba, mejorando la calidad de la inferencia estadística.
