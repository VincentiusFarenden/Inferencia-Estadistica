# **Introducción**

En estadística inferencial, el contraste de hipótesis permite tomar decisiones fundamentadas a partir de información muestral. En muchos contextos reales, el interés no se centra en describir una sola población, sino en **comparar dos poblaciones o dos grupos** con el fin de determinar si existen diferencias significativas entre ellas.

En este laboratorio se aborda el contraste de hipótesis en situaciones de comparación, considerando dos parámetros de interés ampliamente utilizados: la **diferencia de medias poblacionales** y la **diferencia de proporciones poblacionales**. Estos procedimientos permiten evaluar, por ejemplo, si dos grupos presentan rendimientos distintos, si un tratamiento produce efectos diferentes respecto de otro o si la proporción de ocurrencia de un evento varía entre dos poblaciones.

Si bien el tipo de variable y el estadístico de prueba pueden cambiar según el caso, ambos contrastes comparten una **misma estructura inferencial**: formulación de hipótesis nula y alternativa, selección del estadístico de prueba adecuado, cálculo del valor $p$ y toma de decisiones basada en un nivel de significancia previamente definido.

El objetivo de este laboratorio es que comprendas y apliques el contraste de hipótesis para la diferencia de medias y la diferencia de proporciones, interpretando correctamente los resultados obtenidos y relacionándolos con situaciones reales provenientes de ámbitos como la educación, la salud, la industria y las ciencias sociales.

## **Recordatorio clave**

### Hipótesis estadísticas en la comparación de dos poblaciones

En problemas de comparación entre dos poblaciones independientes, el contraste de hipótesis se centra en evaluar si existe una diferencia significativa entre los parámetros de interés.

- **Hipótesis nula $(H_{0})$**: no existe diferencia entre las poblaciones.
- **Hipótesis alternativa $(H_{1})$**: existe una diferencia entre las poblaciones, la cual puede ser **bilateral** o **unilateral**, según el contexto del problema.
<br>

### Contraste para la diferencia de medias poblacionales

**Hipótesis generales:**

$$H_{0}: \ \mu_1 = \mu_2 \quad \text{o} \quad H_{0}:\mu_1 \leq \mu_2 \quad \text{o} \quad H_{0}:\mu_1 \geq \mu_2$$

$$H_{1}: \ \mu_1 \neq \mu_2 \quad \text{o} \quad H_{1}:\mu_1 > \mu_2 \quad \text{o} \quad H_{1}:\mu_1 < \mu_2$$
<br>

#### Estadístico de prueba

En este laboratorio, para contrastar hipótesis sobre la diferencia entre dos medias de muestras independientes, se utilizará la **prueba $t$ de Welch**.

El estadístico de prueba está dado por:

$$t = \frac{\bar{X}_1 - \bar{X}_2}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}$$

con grados de libertad aproximados:

$$gl =
\frac{
\left( \frac{s_1^2}{n_1} + \frac{s_2^2}{n_2} \right)^2
}{
\frac{\left( \frac{s_1^2}{n_1} \right)^2}{n_1 - 1} +
\frac{\left( \frac{s_2^2}{n_2} \right)^2}{n_2 - 1}
}$$

**Nota:** En algunos contextos, cuando se puede asumir igualdad de varianzas poblacionales, puede utilizarse la prueba t de Student. Sin embargo, en este laboratorio trabajaremos únicamente con la prueba t de Welch.
<br>

### Contraste para la diferencia de proporciones poblacionales

**Hipótesis generales:**

$$H_{0}: \ p_1 = p_2 \quad \text{o} \quad H_{0}:p_1 \leq p_2 \quad \text{o} \quad H_{0}:p_1 \geq p_2$$

$$H_{1}: \ p_1 \neq p_2 \quad \text{o} \quad H_{1}:p_1 > p_2 \quad \text{o} \quad H_{1}:p_1 < p_2$$

#### Estadístico de prueba

Las proporciones muestrales se definen como:

$$\hat{p}_1 = \frac{x_1}{n_1},
\qquad
\hat{p}_2 = \frac{x_2}{n_2}$$

Bajo la hipótesis nula, se utiliza la **proporción combinada**:

$$\hat{p} = \frac{x_1 + x_2}{n_1 + n_2}$$

El estadístico de prueba es:

$$z = \frac{\hat{p}_1 - \hat{p}_2}
{\sqrt{\hat{p}(1 - \hat{p})\left(\frac{1}{n_1} + \frac{1}{n_2}\right)}}$$
<br>

### Regla de decisión

Para ambos contrastes:

- Si **valor $p$ $<$ $\alpha$** → se **rechaza $H_{0}$**.
- Si **valor $p$ $≥$ $\alpha$** → **no se rechaza $H_{0}$**.

donde $\alpha$ corresponde al **nivel de significancia** previamente establecido.
<br>

### Errores tipo I y tipo II

En el contraste de hipótesis para la comparación entre dos poblaciones, pueden cometerse dos tipos de errores en la toma de decisiones:

- **Error tipo I $(\alpha)$**: rechazar la hipótesis nula cuando en realidad **no existe diferencia** entre las poblaciones (por ejemplo, $\mu_1 = \mu_2$ o $p_1 = p_2$).
- **Error tipo II $(\beta)$**: no rechazar la hipótesis nula cuando en realidad **sí existe una diferencia** entre las poblaciones (por ejemplo, $\mu_1 \neq \mu_2$ o $p_1 \neq p_2$).

El valor de $\alpha$ corresponde al **nivel de significancia** fijado previamente y representa la probabilidad de cometer un error tipo I. El complemento $1 - \beta$ se denomina **poder de la prueba** y mide la capacidad del contraste para detectar una diferencia real cuando esta existe.

*Nota:* Aumentar el **tamaño muestral** contribuye a disminuir la probabilidad de cometer errores tipo I y tipo II, y a incrementar el poder de la prueba, mejorando la calidad de la inferencia estadística.

# **Actividad 1 — Productividad en dos turnos de trabajo**

Una empresa está interesada en evaluar si existe una diferencia en la productividad promedio (unidades producidas por hora) entre los trabajadores del *turno diurno* y el *turno nocturno*. Para ello, se seleccionaron dos muestras aleatorias independientes de trabajadores de ambos turnos.

Con esta información, la gerencia busca identificar si el horario de trabajo influye en el desempeño y, en caso de haber diferencias, analizar posibles ajustes en la organización de los turnos.
A partir de las muestras, se obtuvo la siguiente información:
- Turno día: $n_{1} = 30$, media = 52, $s_{1} = 5$.  
- Turno noche: $n_{2} = 28$, media = 49, $s_{2} = 6$.

Considerando un nivel de significancia de $\alpha = 0,05$.


a.	Formule la hipótesis nula y la hipótesis alternativa.

**Hipótesis**  
- $H_{0}: \mu_{diurno} = \mu_{nocturno}$  
- $H_{1}: \mu_{diurno} \neq \mu_{nocturno}$

En lenguaje verbal productividad promedio
- $H_{0}:$ productividad promedio turno diurno $=$ productividad promedio turno nocturno
- $H_{1}:$ productividad promedio turno diurno $\neq$ productividad promedio turno nocturno

b.	Calcule el estadístico de prueba.

```python
import numpy as np
from scipy.stats import t

# Datos
n1, x1, s1 = 30, 52, 5
n2, x2, s2 = 28, 49, 6
alpha = 0.05

# Error estándar
se = np.sqrt(s1**2/n1 + s2**2/n2)

# t de Welch (no asume varianzas iguales)
t_turnos = (x1 - x2) / se

# Grados de libertad
gl = (se**4) / (((s1**2/n1)**2)/(n1-1) + ((s2**2/n2)**2)/(n2-1))


print(f"Estadístico de prueba t = {t_turnos:.3f}")
```

c.	Determine el valor $p$.

```python
# Cálculo valor p
p_valor = 2 * (1 - t.cdf(abs(t_turnos), gl))
print(f"Valor p = {p_valor:.4f}")
```

d.	Compare el valor $p$ con el nivel de significancia y establezca la decisión del contraste.

```python
# Comparación y decisión

if p_valor < alpha:
    decision = "Rechazar H0"
else:
    decision = "No rechazar H0"

print(decision)
```

e.	Redacte una conclusión final en el contexto del problema. ¿Existe evidencia suficiente para afirmar que la productividad promedio difiere entre el turno diurno y el turno nocturno?

El contraste de hipótesis arroja un estadístico $t$ = 2,061 con un p-valor = 0,0443.
Dado que el valor $p$ es ligeramente menor que el nivel de significancia habitual $(\alpha = 0,05)$, se rechaza $H_{0}$ y se concluye que sí existen diferencias significativas en la productividad entre los dos turnos. En otras palabras, la evidencia sugiere que el turno diurno y el nocturno no tienen el mismo rendimiento promedio.

# **Actividad 2 — Rendimiento académico en dos grupos**

Un profesor aplicó la misma prueba en dos secciones de la asignatura, pero las secciones tuvieron metodologías de trabajo distintas: el Grupo 1 realizó controles semanales y el Grupo 2 hizo dos controles acumulativos. Aunque el contenido evaluado fue idéntico, el equipo docente quiere saber si estas diferencias operativas se reflejan en el rendimiento promedio.
Los puntajes obtenidos por cada grupo son los siguientes:

- Grupo 1: $n_{1}$ = 25, media = 68, $s_{1}$ = 8  
- Grupo 2: $n_{2}$ = 22, media = 72, $s_{2}$ = 7  

Realiza un test de hipótesis para diferencia de medias con un nivel de significancia $\alpha = 0.05$.  


a.	Formule la hipótesis nula y la hipótesis alternativa.

**Hipótesis**

- $H_{0} : \mu_{grupo1} = \mu_{grupo2}$

- $H_{1} : \mu_{grupo1} \neq \mu_{grupo2}$

En lenguaje verbal

- $H_{0} :$ Rendimiento promedio grupo 1 $=$ Rendimiento promedio grupo 2

- $H_{1} :$ Rendimiento promedio grupo 1 $\neq$ Rendimiento promedio grupo 2

b.	Calcule el estadístico de prueba.

```python
import numpy as np
from scipy.stats import t

# Datos
n1, x1, s1 = 25, 68, 8
n2, x2, s2 = 22, 72, 7
alpha = 0.05

# Error estándar
se = np.sqrt(s1**2/n1 + s2**2/n2)

# t de Welch (no asume varianzas iguales)
t_rendimiento = (x1 - x2) / se

# Grados de libertad
gl = (se**4) / (((s1**2/n1)**2)/(n1-1) + ((s2**2/n2)**2)/(n2-1))

print(f"Estadístico de prueba t = {t_rendimiento:.3f}")
```

c.	Determine el valor $p$.

```python
# Cálculo valor p
p_valor = 2 * (1 - t.cdf(abs(t_rendimiento), gl))
print(f"Valor p = {p_valor:.4f}")
```

d.	Compare el valor $p$ con el nivel de significancia y establezca la decisión del contraste.

```python
# Comparación y decisión

if p_valor < alpha:
    decision = "Rechazar H0"
else:
    decision = "No rechazar H0"

print(decision)
```

e.	Redacte una conclusión en el contexto del problema. ¿Existe evidencia suficiente para afirmar que el rendimiento promedio difiere entre los grupos que trabajaron con metodologías distintas?

Con $\alpha = 0,05$, el valor $p$ = 0,0742 es mayor que 0,05. por lo tanto no se rechaza $H_{0}$. Aunque el grupo 2 obtuvo un puntaje promedio mayor que el grupo 1, la evidencia no es suficiente al 5% para afirmar que los promedios poblacionales difieran.

# **Actividad 3 — Tratamiento médico vs placebo**

En un ensayo clínico controlado, dos grupos de pacientes con hipertensión leve reciben intervenciones distintas durante dos semanas. El grupo tratamiento recibe un nuevo fármaco, mientras que el grupo placebo recibe una pastilla inerte con el mismo aspecto para mantener el cegamiento. Los médicos quieren evaluar si el tratamiento logra una mayor reducción promedio de la presión arterial que el placebo.

Al finalizar las dos semanas, se obtuvo la siguiente información resumen sobre la reducción de presión arterial, medida en mmHg:
- Tratamiento: $n_{1}$ = 20, media = 12 mmHg, $s_{1}$ = 4  
- Placebo: $n_{2}$ = 18, media = 9 mmHg, $s_{2}$ = 3  

Determina si el nuevo tratamiento logra una mayor reducción promedio que el placebo. Utiliza un nivel de significancia del 1%.

1. Formule las hipótesis

**Hipótesis**

- $H_{0} : \mu_{tratamiento} \leq \mu_{placebo}$

- $H_{1} : \mu_{tratamiento} > \mu_{placebo}$

En lenguaje verbal
- $H_{0} :$ Reducción promedio grupo tratamiento $\leq$ Reducción promedio grupo placebo

- $H_{1} :$ Reducción promedio grupo tratamiento $>$ Reducción promedio grupo placebo


2. Construya el contraste de hipótesis

```python
import scipy.stats as st

# Datos
n1, x1, s1 = 20, 12, 4
n2, x2, s2 = 18, 9, 3
alpha = 0.01

# Error estándar
se = np.sqrt(s1**2/n1 + s2**2/n2)

# t de Welch (no asume varianzas iguales)
t = (x1 - x2) / se

# Grados de libertad
gl = (se**4) / (((s1**2/n1)**2)/(n1-1) + ((s2**2/n2)**2)/(n2-1))

# Cálculo valor p
p_valor = 1 - st.t.cdf(t, gl)

print(f"t = {t:.3f}, p-valor = {p_valor:.4f}")
```

```python
# Comparación y decisión

if p_valor < alpha:
    decision = "Rechazar H0"
else:
    decision = "No rechazar H0"

print(decision)
```

3. Interprete los resultados obtenidos en el contexto del problema

Con $\alpha = 0,01$, el valor $p$ = 0,0063 es menor que 0,01. por lo tanto, se rechaza $H_{0}$. Hay evidencia estadística de que el tratamiento reduce en promedio más la presión arterial que el placebo.

# **Actividad 4 — Eficiencia energética en motores eléctricos**

Una empresa de manufactura reemplazó recientemente los motores de una de sus líneas de producción por un nuevo modelo con mayor eficiencia energética. El departamento de ingeniería desea evaluar si el consumo promedio de energía eléctrica, medido en kWh por hora de operación, ha disminuido con el nuevo modelo en comparación con el modelo anterior.

Para ello, se dispone de un conjunto de datos que contiene mediciones de consumo energético correspondientes a motores del modelo antiguo y del nuevo modelo.

Descargue el set de datos **<a href="https://docs.google.com/spreadsheets/d/154B6V7Ug2Kcd_4fPGTPp1--K-VjinZAV/export?format=xlsx">aquí</a>.**

A partir de esta información, realice un contraste de hipótesis para determinar si el consumo promedio del nuevo modelo es menor que el del modelo antiguo, utilizando un nivel de significancia del 5%. Para ello:


a.	Formule las hipótesis del problema.

**Hipótesis**

- $H_{0} : \mu_{nuevo} \geq \mu_{antiguo}$

- $H_{1} : \mu_{nuevo} < \mu_{antiguo}$

En lenguaje verbal
- $H_{0} :$ Consumo promedio nuevo $\geq$ Consumo promedio antiguo

- $H_{1} :$ Consumo promedio nuevo $<$ Consumo promedio antiguo


b.	Cargue la base de datos al entorno de trabajo.

```python
import pandas as pd
df = pd.read_excel("G5_A4_consumo_motores.xlsx")
df.head(3)
```

c.	Calcule los estadísticos necesarios para realizar el contraste de hipótesis.

```python
import numpy as np
from scipy.stats import t
import scipy.stats as st

# Datos
# Extraer muestras desde el df
x1 = df["antiguo_kwh"]  # modelo antiguo
x2 = df["nuevo_kwh"]    # modelo nuevo

# Estadísticos muestrales necesarios ---
n1, n2 = len(x1), len(x2)
x1_bar, x2_bar = x1.mean(), x2.mean()
s1, s2 = x1.std(ddof=1), x2.std(ddof=1)  # sd muestral (ddof=1)

print(f"Antiguo -> n={n1}, media={x1_bar:.1f}, s={s1:.1f}")
print(f"Nuevo   -> n={n2}, media={x2_bar:.1f}, s={s2:.1f}")
```

d.	Realice el contraste para la diferencia de medias.

```python
# Nivel de significancia
alpha = 0.05

# Error estándar
se = np.sqrt(s1**2/n1 + s2**2/n2)

# t de Welch (no asume varianzas iguales)
t_motores = (x2_bar - x1_bar) / se

# Grados de libertad
gl = (se**4) / (((s1**2/n1)**2)/(n1-1) + ((s2**2/n2)**2)/(n2-1))

print(f"Estadístico de prueba t = {t_motores:.3f}")
```

```python
# Cálculo valor p
p_valor = st.t.cdf(t_motores, df=gl)
print(f"Valor p = {p_valor:.4f}")
```

```python
# Comparación y decisión
if p_valor < alpha:
    decision = "Rechazar H0"
else:
    decision = "No rechazar H0"

print(decision)
```

e.	Interprete los resultados obtenidos en el contexto del problema.

Con $\alpha = 0,05$, el valor $p$ = 0,0471 es menor que 0,05. por lo tanto, se rechaza $H_{0}$. Existe evidencia estadística suficiente al 5 % para afirmar que el nuevo modelo consume menos energía en promedio que el modelo antiguo.

# **Contraste de hipótesis para la diferencia de proporciones**

## **Transición: de la diferencia de medias a la diferencia de proporciones**

Hasta este punto, has trabajado el contraste de hipótesis para la **diferencia de medias poblacionales**, el cual se utiliza cuando el interés está en comparar el comportamiento promedio de una variable cuantitativa entre dos poblaciones independientes.

En muchos contextos reales, sin embargo, la comparación no se centra en valores promedio, sino en la **proporción de individuos que presentan una determinada característica** en cada población, como la ocurrencia de un evento, la aprobación de un curso o la adopción de una conducta específica. En estos casos, la variable de interés es de tipo categórica dicotómica y el parámetro que se compara es la **diferencia de proporciones poblacionales**.

# **Actividad 6 — Efectividad de clases híbridas**

Un Instituto Profesional detectó que ciertas asignaturas presentan altas tasas de reprobación y decidió probar una estrategia de clases híbridas (combinación de sesiones presenciales con actividades en línea sincrónicas/asíncronas) para mejorar el aprendizaje y la retención.

En un semestre se trabajó así:
* Grupo híbrido: 120 estudiantes inscritos, de los cuales 78 aprobaron.
* Grupo presencial tradicional: 150 estudiantes inscritos, de los cuales 85 aprobaron.

Como la implementación de clases híbridas implica cambios en la planificación docente, uso de plataformas y capacitación del profesorado, la universidad quiere saber si realmente hay diferencia en la tasa de aprobación entre ambas modalidades antes de escalar el modelo a más asignaturas.

A un nivel de significancia del 5 %, ¿la proporción de estudiantes que aprueban es distinta entre la modalidad híbrida y la presencial?

a.	Formule la hipótesis nula y la hipótesis alternativa.

**Hipótesis**

- $H_{0} : p_{híbrido} = p_{presencial}$
- $H_{1} : p_{híbrido} \neq p_{presencial}$

El lenguaje verbal
- $H_{0} :$ Proporción de estudiantes formato híbrido $=$ Proporción de estudiantes formato presencial
- $H_{1} :$ Proporción de estudiantes formato híbrido $\neq$ Proporción de estudiantes formato presencial

b.	Calcule el estadístico de prueba.

```python
import numpy as np
from scipy.stats import norm
import scipy.stats as st

# Datos
n1, x1 = 120, 78 # Grupo híbrido
n2, x2 = 150, 85 # Grupo presencial
alpha = 0.05

# Proporciones muestrales
p1 = x1 / n1
p2 = x2 / n2

# Proporción combinada
p = (x1 + x2) / (n1 + n2)

# Estadístico de Prueba
z_clases = (p1 - p2) / np.sqrt(p*(1-p)*(1/n1 + 1/n2))

print(f"Estadístico de prueba z = {z_clases:.3f}")
```

c.	Determine el valor $p$.

```python
# Cálculo valor p
p_valor = 2 * (1 - st.norm.cdf(abs(z_clases)))

print(f"Valor p = {p_valor:.4f}")
```

d.	Compare el valor $p$ con el nivel de significancia y establezca la decisión del contraste.

```python
# Comparación y decisión
if p_valor < alpha:
    decision = "Rechazar H0"
else:
    decision = "No rechazar H0"

print(decision)
```

e.	Interprete los resultados obtenidos en el contexto del problema.

Con un nivel de significancia $\alpha = 0.05$, y un valor $p$ = 0,1642, dado que el valor $p$ es mayor que el valor de $\alpha$, no se rechaza $H_{0}$. Con los datos disponibles y usando un nivel de significancia del 5 %, no se puede afirmar que la modalidad híbrida tenga una tasa de aprobación diferente a la presencial.

¿Hay evidencia al 1 % para concluir que el Centro A tiene una proporción mayor de vacunación que el Centro B?

**Hipótesis**

- $H_{0} : p_{centroA} \leq p_{centroB}$
- $H_{1} : p_{centroA} > p_{centroB}$

En lenguaje verbal
- $H_{0} :$ Proporción vacunados centro A $\leq$ Proporción vacunados centro B
- $H_{1} :$ Proporción vacunados centro A $>$ Proporción vacunados centro B

**Hipótesis**

Un equipo de desarrollo de software está evaluando el rendimiento de dos versiones de un mismo microservicio que procesa solicitudes de usuarios. La métrica clave es el tiempo de respuesta promedio (latencia en milisegundos).

- La versión A corresponde al servicio actual en producción.
- La versión B corresponde a una nueva versión con cambios en la arquitectura.

Antes de decidir si se mantiene la versión actual o se libera la nueva versión, el equipo desea comprobar si la latencia promedio del servicio actual es menor que la de la nueva versión, ya que un aumento en la latencia podría afectar negativamente la experiencia de usuario.

Para ello, se comparan dos muestras independientes de tiempos de respuesta, obtenidas bajo condiciones equivalentes de carga.

- Versión A: $n_{1}$ = 30, media = 198 ms, $s_{1}$ = 12  
- Versión B: $n_{2}$ = 30, media = 205 ms, $s_{2}$ = 10  

Con un nivel de significancia del 5 %, determine si existe evidencia estadísticamente suficiente para afirmar que la latencia promedio de la versión A es menor que la de la versión B. Luego de realizar el contraste de hipótesis correspondiente, ¿qué decisión debería tomar el equipo respecto del despliegue de la nueva versión?

1. Formule las hipótesis

**Hipótesis**

- $H_{0} : \mu_{versionA} >= \mu_{versionB}$

- $H_{1} : \mu_{versionA} < \mu_{versionB}$

En lenguaje verbal
- $H_{0} :$ Latencia promedio versión A $\geq$ Latencia promedio versión B

- $H_{1} :$ Latencia promedio versión A $<$ Latencia promedio versión B

2. Construya el constraste de hipótesis

```python
import numpy as np
from scipy import stats

# Datos
n1, x1, s1 = 30, 198, 12   # Versión A
n2, x2, s2 = 30, 205, 10   # Versión B
alpha = 0.05

# Error estándar
se = np.sqrt(s1**2/n1 + s2**2/n2)

# t de Welch (no asume varianzas iguales)
t_version = (x1 - x2) / se

# Grados de libertad
gl = (se**4) / (((s1**2/n1)**2)/(n1-1) + ((s2**2/n2)**2)/(n2-1))

# Cálculo valor p
p_valor = stats.t.cdf(t_version, df=gl)

# Comparación y decisión
if p_valor < alpha:
    decision = "Rechazar H0"
else:
    decision = "No rechazar H0"

print(f"Estadístico de prueba t = {t_version:.3f}, Valor p = {p_valor:.4f}, Decisión = {decision}")
```

3. Interprete el resultado

Con $\alpha = 0,05$, el valor $p$ = 0,0086 es menor que 0,05. por lo tanto, se rechaza $H_{0}$, existe evidencia estadísticamente significativa para afirmar que la latencia promedio de la versión actual (A) es menor que la de la nueva versión (B), bajo las condiciones analizadas.

**Hipótesis**

- $H_{0} : p_{2022-2024} \leq p_{2015-2017}$
- $H_{1} : p_{2022-2024} > p_{2015-2017}$

En lenguaje verbal
- $H_{0} :$ Proporción de mujeres cohorte 2022-2024 $\leq$ Proporción de mujeres cohorte 2015-2017
- $H_{1} :$ Proporción de mujeres cohorte 2022-2024 $>$ Proporción de mujeres cohorte 2015-2017

2. Construya el contraste de hipótesis

# **Actividad 13 - Control de calidad: pureza del ácido en dos plantas**

Una empresa química desea comparar la pureza del ácido sulfúrico producido en dos plantas, ya que diferencias sistemáticas entre ellas podrían afectar los estándares de calidad del proceso.

Se registraron los siguientes porcentajes de pureza del ácido en muestras obtenidas desde cada planta:

*	Planta A: 96.8, 95.8, 96.2, 96.5, 96.4, 96.7, 96.3, 96.0, 97.1, 96.5
*	Planta B: 95.6, 96.0, 97.8, 98.4, 97.6, 98.2, 96.9, 96.8

Con un nivel de significancia del 1 %, determine si existe evidencia estadísticamente suficiente para afirmar que los porcentajes promedios de pureza del ácido difieren entre la Planta A y la Planta B.
Luego de realizar el contraste de hipótesis correspondiente, ¿qué conclusión operativa podría extraer la empresa respecto del control de calidad entre ambas plantas?

1. Formule las hipótesis

**Hipótesis**

- $H_{0} : \mu_{plantaA} = \mu_{plantaB}$

- $H_{1} : \mu_{plantaA} \neq \mu_{plantaB}$

En lenguaje verbal
- $H_{0} :$ Pureza promedio planta A $=$ Pureza promedio planta B

- $H_{1} :$ Pureza promedio planta A $\neq$ Pureza promedio planta B

2. Construya el constraste de hipótesis

```python
import numpy as np
from scipy.stats import t
import scipy.stats as st

# Datos
A = np.array([96.8, 95.8, 96.2, 96.5, 96.4, 96.7, 96.3, 96.0, 97.1, 96.5])
B = np.array([95.6, 96.0, 97.8, 98.4, 97.6, 98.2, 96.9, 96.8])
alpha = 0.01

# Estadísticos muestrales
n1, n2 = len(A), len(B)
x1, x2 = A.mean(), B.mean()
s1, s2 = A.std(ddof=1), B.std(ddof=1)

# Error estándar
se = np.sqrt(s1**2/n1 + s2**2/n2)

# t de Welch (no asume varianzas iguales)
t_pureza = (x1 - x2) / se

# Grados de libertad
gl = (se**4) / (((s1**2/n1)**2)/(n1-1) + ((s2**2/n2)**2)/(n2-1))

# Cálculo valor p
p_valor = 2 * (1 - t.cdf(abs(t_pureza), gl))

print(f"nA = {n1}, nB = {n2}")
print(f"mediaA = {x1:.3f}, mediaB = {x2:.3f}")
print(f"sA = {s1:.3f}, sB = {s2:.3f}")
print()
print(f"t = {t_pureza:.3f}, p-valor = {p_valor:.4f}")
```

```python
# Comparación y decisión
if p_valor < alpha:
    decision = "Rechazar H0"
else:
    decision = "No rechazar H0"

print(decision)
```

3. Interprete el resultado

Con $\alpha = 0,01$, el valor $p$ = 0,0864 es mayor que 0,01. por lo tanto, no se rechaza $H_{0}$. No hay evidencia suficiente al 1% para afirmar que los promedios de pureza del ácido sean diferentes entre la planta A y planta B.
