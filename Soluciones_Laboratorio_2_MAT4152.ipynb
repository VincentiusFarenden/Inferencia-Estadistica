## **Recordatorio clave**

| Situación                                    | Fórmula para el IC de la media |
|---------------------------------------------|--------------------------------|
| **σ poblacional desconocida**                | $ \bar{x} \pm t_{\alpha/2, \, n-1} \cdot \frac{s}{\sqrt{n}} $ |
| **σ poblacional conocida**                   | $ \bar{x} \pm z_{\alpha/2} \cdot \frac{\sigma}{\sqrt{n}} $ |


- **$n$**: tamaño de la muestra  
- **$\bar{x}$**: media muestral  
- **$s$**: desviación estándar muestral  
- **$\sigma$**: desviación estándar poblacional conocida  
- **$z$** o **$t$**: valor crítico según nivel de confianza elegido

💡 **Decisión rápida:**  
- Si la desviación estándar poblacional ($\sigma$) es desconocida y la muestra es pequeña ($n < 30$), usar distribución $t$ de Student.

- Si la desviación estándar poblacional ($\sigma$) es conocida o la muestra es grande ($n ≥ 30$), usar distribución $z$.

# **Actividad 1 — Estimación de la temperatura promedio de un laboratorio**

En un laboratorio de biotecnología que manipula microorganismos sensibles, mantener la temperatura ambiente dentro de un rango controlado es crucial para la estabilidad de los cultivos y la seguridad del personal. Durante una jornada de trabajo matutina se registraron 25 mediciones de temperatura en distintos puntos del laboratorio, ya que el sistema de climatización había presentado variaciones y se requería verificar si las condiciones se mantenían estables.

A continuación, se muestran las mediciones de temperatura (°C) registradas:

<center>

|  |  |  |  |  |
|------|------|------|------|------|
| 22.1 | 22.3 | 22.7 | 21.9 | 22.5 |
| 22.8 | 22.0 | 22.4 | 21.8 | 22.3 |
| 22.6 | 22.5 | 22.1 | 22.4 | 22.7 |
| 22.8 | 22.0 | 22.2 | 22.1 | 22.5 |
| 22.3 | 22.7 | 22.6 | 22.4 | 22.5 |

1. Calcula la *media* y la *desviación estándar muestral* de los datos de temperatura.  

```python
import numpy as np
import scipy.stats as st

temperaturas = [22.1, 22.3, 22.7, 21.9, 22.5,
                22.8, 22.0, 22.4, 21.8, 22.3,
                22.6, 22.5, 22.1, 22.4, 22.7,
                22.8, 22.0, 22.2, 22.1, 22.5,
                22.3, 22.7, 22.6, 22.4, 22.5]

# Cálculos
media = np.mean(temperaturas)
s = np.std(temperaturas, ddof=1)
print(f"Media muestral: {media:.2f} °C")
print(f"Desviación estándar muestral: {s:.2f} °C")
```

2. Construye un *intervalo de confianza al 95%* para la media poblacional usando la distribución *t* de Student.

```python
n = len(temperaturas)

ic = st.t.interval(
    confidence=0.95,   # Nivel de confianza
    df=n-1,            # Grados de libertad
    loc=media,         # Media muestral
    scale=s/np.sqrt(n) # Error estándar: desviación estándar dividida por la raíz del tamaño muestral
)

print(f"Intervalo de confianza 95%: {ic[0]:.2f} °C a {ic[1]:.2f} °C")
```

3. ¿Qué significa que el intervalo de confianza tenga un nivel de confianza del 95%?

**Interpretación**

Esto significa que, si se repitiera este procedimiento de medición muchas veces, aproximadamente el 95% de los intervalos construidos con este método contendrían la verdadera temperatura promedio del laboratorio.

4. Interpreta el intervalo de confianza encontrado en un lenguaje comprensible para el responsable del laboratorio.

**Interpretación**

La temperatura promedio registrada fue de aproximadamente 22,37 °C.
El intervalo de confianza al 95% para la temperatura media del laboratorio se encuentra entre 22,25 °C y 22,49 °C. En términos prácticos, podemos comunicar al responsable que la temperatura ambiente se mantiene estable alrededor de los 22 °C, con un rango de estimación estrecho, indicando un buen control térmico.


# **Actividad 2 — Resultados académicos en educación**

En un instituto profesional se busca estimar el promedio final de notas en una asignatura clave con una matrícula de 800 estudiantes. Para tener una primera aproximación sin procesar todos los registros, se tomó una muestra aleatoria de 20 calificaciones finales (escala de 1,0 a 7,0). De estos datos desconoce la desviación estándar poblacional.

Las calificaciones recolectadas son:

<center>

|  |  |  |  |  |  |  |  |  |  |
|-----|-----|-----|-----|-----|-----|-----|-----|-----|-----|
| 5.1 | 5.4 | 4.8 | 5.2 | 4.9 | 5.5 | 5.0 | 4.7 | 5.3 | 5.1 |
| 5.2 | 5.0 | 5.3 | 5.5 | 4.9 | 5.4 | 5.2 | 5.0 | 5.6 | 5.3 |



1. Calcula la *media* y la *desviación estándar muestral*.

```python
import numpy as np
import scipy.stats as st

notas = [5.1,5.4,4.8,5.2,4.9,5.5,5.0,4.7,5.3,5.1,
         5.2,5.0,5.3,5.5,4.9,5.4,5.2,5.0,5.6,5.3]

media = np.mean(notas)
s = np.std(notas, ddof=1)
print(f"Media muestral: {media:.2f}")
print(f"Desviación estándar muestral: {s:.2f}")
```

2. Antes de construir el intervalo, justifica brevemente si usarías $z$ o $t$ y por qué. ¿Qué información sobre la desviación estándar poblacional ($\sigma$) respalda esta decisión?

En este caso se debe usar la distribución t de Student, porque no se conoce la desviación estándar poblacional ($\sigma$) y solo se dispone de una muestra pequeña ($n = 20$).
La distribución $t$ se utiliza justamente cuando trabajamos con una desviación estándar muestral ($s$) como estimación de $\sigma$.

3. Construye un *intervalo de confianza al 95%* para la media poblacional.

```python
n = len(notas)
ic = st.t.interval(0.95, df=n-1, loc=media, scale=s/np.sqrt(n))

print(f"IC 95%: {ic[0]:.2f} a {ic[1]:.2f}")
```

**Interpretación**

La nota promedio estimada para la asignatura es de 5,17.
Con un nivel de confianza al 95%, la media poblacional de las notas se encuentra entre 5,05 y 5,29.

Este rango indica que, si repitiéramos el muestreo muchas veces, aproximadamente el 95 % de los intervalos construidos con este método contendrían la verdadera media de la población de estudiantes.

# **Actividad 3 — Efecto del tamaño de muestra y del nivel de confianza**

En esta actividad explorarás cómo cambian los intervalos de confianza cuando varía el tamaño de la muestra y el nivel de confianza.

Trabaja con los resultados de un examen nacional de matemáticas con media 500 puntos y desviación estándar 100 puntos, asumiendo que los puntajes se distribuyen normalmente.

### **Parte A: Efecto del tamaño muestral**

1. Genera un conjunto de 100 puntajes simulados que sigan una distribución normal

```python
import numpy as np
import scipy.stats as st

# Datos simulados
np.random.seed(42)
puntajes = np.random.normal(loc=500, scale=100, size=100)
```

2. Calcula los intervalos de confianza al $95\%$ para muestras de tamaños $n=10,30,100$. Luego, determina la amplitud de cada intervalo.

```python
# Cálculo para n = 10
n = 10
muestra_10 = np.random.choice(puntajes, size=n, replace=False)
media_10 = np.mean(muestra_10)
s_10 = np.std(muestra_10, ddof=1)
ic_10 = st.t.interval(0.95, df=n-1, loc=media_10, scale=s_10/np.sqrt(n))
amplitud_10 = ic_10[1] - ic_10[0]

print(f"n = {n}, Media = {media_10:.2f}")
print(f"IC 95%: {ic_10[0]:.2f} a {ic_10[1]:.2f}")
print(f"Amplitud: {amplitud_10:.2f}\n")
```

```python
# Cálculo para n = 30
n = 30
muestra_30 = np.random.choice(puntajes, size=n, replace=False)
media_30 = np.mean(muestra_30)
s_30 = np.std(muestra_30, ddof=1)
ic_30 = st.t.interval(0.95, df=n-1, loc=media_30, scale=s_30/np.sqrt(n))
amplitud_30 = ic_30[1] - ic_30[0]

print(f"n = {n}, Media = {media_30:.2f}")
print(f"IC 95%: {ic_30[0]:.2f} a {ic_30[1]:.2f}")
print(f"Amplitud: {amplitud_30:.2f}\n")
```

```python
# Cálculo para n = 100
n = 100
muestra_100 = np.random.choice(puntajes, size=n, replace=False)
media_100 = np.mean(muestra_100)
s_100 = np.std(muestra_100, ddof=1)
ic_100 = st.t.interval(0.95, df=n-1, loc=media_100, scale=s_100/np.sqrt(n))
amplitud_100 = ic_100[1] - ic_100[0]

print(f"n = {n}, Media = {media_100:.2f}")
print(f"IC 95%: {ic_100[0]:.2f} a {ic_100[1]:.2f}")
print(f"Amplitud: {amplitud_100:.2f}")
```

**Reflexiona:**
* ¿Qué ocurre con el intervalo al aumentar el tamaño de la muestra?
* ¿Qué puedes concluir sobre la precisión de la estimación?

**Interpretación**

* ¿Qué ocurre con el intervalo al aumentar el tamaño de la muestra?

El intervalo de confianza se vuelve más estrecho cuando aumenta el tamaño de la muestra.

* ¿Qué puedes concluir sobre la precisión de la estimación?

Al tener una muestra más grande, la estimación de la media es más precisa y hay menos incertidumbre sobre el valor real de la población.

### **Parte B: Efecto del nivel de confianza**

3. Usando una muestra fija de tamaño $n = 30$, calcula los intervalos al $90\%$, $95\%$ y $99\%$ de confianza y su amplitud.

```python
# Datos simulados
np.random.seed(42)
puntajes = np.random.normal(loc=500, scale=100, size=100)
```

```python
# Tamaño de muestra fijo
n = 30
muestra_30 = np.random.choice(puntajes, size=n, replace=False)
media_30 = np.mean(muestra_30)
s_30 = np.std(muestra_30, ddof=1)
```

```python
# IC 90%
ic_90 = st.t.interval(0.90, df=n-1, loc=media_30, scale=s_30/np.sqrt(n))
amp_90 = ic_90[1] - ic_90[0]

print("Nivel de confianza: 90%")
print(f"IC: {ic_90[0]:.2f} a {ic_90[1]:.2f}")
print(f"Amplitud: {amp_90:.2f}\n")
```

```python
# IC 95%
ic_95 = st.t.interval(0.95, df=n-1, loc=media_30, scale=s_30/np.sqrt(n))
amp_95 = ic_95[1] - ic_95[0]

print("Nivel de confianza: 95%")
print(f"IC: {ic_95[0]:.2f} a {ic_95[1]:.2f}")
print(f"Amplitud: {amp_95:.2f}\n")
```

```python
# IC 99%
ic_99 = st.t.interval(0.99, df=n-1, loc=media_30, scale=s_30/np.sqrt(n))
amp_99 = ic_99[1] - ic_99[0]

print("Nivel de confianza: 99%")
print(f"IC: {ic_99[0]:.2f} a {ic_99[1]:.2f}")
print(f"Amplitud: {amp_99:.2f}")
```

**Reflexiona**
* ¿Qué ocurre con el intervalo al aumentar el nivel de confianza?
* ¿Qué implicancia tiene esto en términos de *“seguridad”* y *“precisión”*?

**Interpretación**

* ¿Qué ocurre con el intervalo al aumentar el nivel de confianza?

Al aumentar el nivel de confianza, el intervalo se hace más amplio. Esto ocurre porque necesitamos cubrir un rango mayor de valores para estar más seguros de que la media real está dentro del intervalo.

* ¿Qué implicancia tiene esto en términos de “seguridad” y “precisión”?

Un intervalo más amplio entrega mayor seguridad (más confianza en que contiene la media real), pero menor precisión, ya que el rango de valores posibles es más grande. En cambio, un intervalo más estrecho es más preciso, pero con menor seguridad.

# **Actividad 4 — Satisfacción de clientes en un comercio electrónico**

Una empresa de venta en línea desea evaluar el nivel de satisfacción de sus clientes para mejorar su servicio postventa. Se tomó una muestra aleatoria de 50 valoraciones de clientes recientes, cada una en una escala de 1 a 10, donde 10 es “muy satisfecho”. Estudios previos de la plataforma sugieren que la desviación estándar poblacional puede asumirse cercana a 1,2 puntos.

En el dataset se presentan los puntajes de satisfacción de la muestra tomada, descargue sus datos **<a href="https://docs.google.com/spreadsheets/d/1Y5-tGT4tJ6xh7g_HPy6yTNgQuP0-IX1w/export?format=xlsx">aquí</a>**:



1. Calcula la *media muestral* de satisfacción.

```python
import pandas as pd
import numpy as np
import scipy.stats as st

valoraciones = pd.read_excel("G2_A4_valoraciones_clientes.xlsx")

media = np.mean(valoraciones)
print(f"Media muestral: {media:.2f}")
```

**Interpretación**
* La satisfacción promedio estimada es de aproximadamente **8,37 puntos** sobre 10.

2. ¿Qué distribución, $z$ o $t$, correspondería utilizar si tuvieses que estimar la media poblacional de la satisfacción de los clientes mediante un intervalo de confianza? Justifica tu respuesta.

Como se conoce la desviación estándar poblacional ($\sigma = 1,2$), se debe utilizar la distribución z, ya que esta se aplica cuando los parámetros poblacionales son conocidos o estimados con alta precisión. 3. Construye un intervalo de confianza al 90% para la media. ¿Qué indica este intervalo sobre el nivel de satisfacción de los clientes?

En otras palabras, la mayoría de los clientes se sienten muy satisfechos, ya que los valores del intervalo son altos dentro de la escala de 1 a 10.

4. Calcula el intervalo al 99% y compara los resultados obtenidos con los del inciso anterior. ¿Qué ocurre con la amplitud del intervalo al aumentar el nivel de confianza?

```python
# Intervalo al 99%
ic_99 = st.norm.interval(0.99, loc=media, scale=sigma/np.sqrt(n))

print(f"IC 99%: {ic_99[0]:.2f} a {ic_99[1]:.2f}")
```

**Interpretación**

El intervalo al 99% (7,93 a 8,81) es más amplio que el del 90%, lo que significa que tenemos más seguridad de incluir la media verdadera, pero menos precisión en la estimación.

**Reflexión**

Esto significa que para tener más confianza (99% en lugar de 90%), el intervalo debe ser más ancho y menos preciso. Para la empresa, un IC al 90% es razonable si se busca tomar decisiones rápidas y el riesgo de error es aceptable; un IC al 99% sería preferible para reportes ejecutivos o decisiones críticas donde se necesita mayor certeza.

# **Actividad 6 - Vida útil de ampolletas**

Un fabricante anuncia que sus ampolletas duran en promedio 800 horas, con una desviación estándar conocida de 40 horas.

Se toma una muestra de 25 ampolletas, cuya vida promedio observada es de 780 horas.

Asumiendo que la duración de las ampolletas se distribuye normalmente:

1.	Calcula un intervalo de confianza del 95% para la media poblacional.




```python
# Estadísticos básicos
media = 780
sigma = 40
n = 25

# Intervalo de confianza
ic6_95 = st.norm.interval(0.95, loc=media, scale=sigma/np.sqrt(n))

print(f"IC 95%: {ic6_95[0]:.2f} a {ic6_95[1]:.2f} horas")
```
