# **Introducción**

En estadística inferencial, el contraste de hipótesis es una herramienta fundamental que permite tomar decisiones sobre una población a partir de información obtenida desde una muestra. Este procedimiento se basa en la formulación de una hipótesis nula, que representa una afirmación inicial sobre un parámetro poblacional, y una hipótesis alternativa, que plantea una posible discrepancia respecto de dicha afirmación.

A lo largo de este laboratorio se abordará el contraste de hipótesis en dos situaciones ampliamente utilizadas en contextos reales: el contraste para la **media poblacional** y el contraste para la **proporción poblacional**. Si bien ambos procedimientos se aplican a distintos tipos de variables, comparten una misma lógica inferencial: definir hipótesis, seleccionar un estadístico de prueba adecuado, establecer una regla de decisión y emitir una conclusión fundamentada en el nivel de significancia.

El objetivo de este laboratorio es que comprendas y apliques el proceso de contraste de hipótesis de manera estructurada, interpretando correctamente los resultados obtenidos y relacionándolos con situaciones prácticas provenientes de ámbitos como la educación, la salud, la industria y la tecnología.

## **Recordatorio clave**

### Hipótesis estadísticas
- **Hipótesis nula $(H_{0})$**: afirmación inicial que se somete a contraste. Representa ausencia de efecto o igualdad con un valor hipotético.
- **Hipótesis alternativa $(H_{1})$**: plantea la existencia de una diferencia, efecto o cambio respecto de $H_{0}$. Puede ser unilateral o bilateral, según el contexto del problema.
<br>

### Estadístico de prueba

El estadístico de prueba depende del **parámetro de interés** y de la información disponible:

**a) Contraste para la media poblacional**
- Si la desviación estándar poblacional es **conocida** → **prueba $z$**.
- Si la desviación estándar poblacional es **desconocida** → **prueba $t$ de Student**.

$$
z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}
\qquad
t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
$$

**b) Contraste para la proporción poblacional**
- Se utiliza la **distribución normal ($z$)**.

$$
z = \frac{p - p_0}{\sqrt{\frac{p_0(1 - p_0)}{n}}}
$$
<br>


### Valor $p$
El **valor $p$** es la probabilidad de obtener un estadístico de prueba igual o más extremo que el observado en la muestra, bajo el supuesto de que la hipótesis nula $(H_0)$ es verdadera.
* Un valor $p$ pequeño indica que lo observado es muy improbable si $H_0$ fuera cierta (evidencia fuerte en contra de $H_0$).
* Un valor $p$ grande indica que lo observado es compatible con $H_0$ (poca evidencia en contra).
<br>

### Regla de decisión
Para concluir, se compara el **valor $p$** con el **nivel de significancia $(\alpha)$**
- Si **valor $p < \alpha$** → se **rechaza $H_{0}$**.
- Si **valor $p ≥ \alpha$** → **no se rechaza $H_{0}$**.

<br>

*Nota:* Al realizar pruebas unilaterales, verifica si la función utilizada entrega el valor p de una o dos colas. Si es bilateral y la prueba es unilateral, se debe ajustar el valor.
<br>


### Tipos de prueba según la hipótesis alternativa

| Tipo de prueba | Hipótesis alternativa | Interpretación |
|---------------|----------------------|----------------|
| Unilateral derecha | $H_{1}: \ \theta > \theta_0$ | Aumento del parámetro |
| Unilateral izquierda | $H_{1}: \ \theta < \theta_0$ | Disminución del parámetro |
| Bilateral | $H_{1}: \ \theta \neq \theta_0$ | Cualquier cambio |

<br>


### Errores tipo I y tipo II
- **Error tipo I $(\alpha)$**: rechazar $H_{0}$ cuando en realidad es verdadera.
- **Error tipo II $(\beta)$**: no rechazar $H_{0}$ cuando en realidad es falsa.
- **$1 - \beta$** corresponde al **poder de la prueba**, es decir, la capacidad de detectar un efecto real.

*Nota:* Reducir $\alpha$ disminuye la probabilidad de error tipo I, pero generalmente aumenta $\beta$. Aumentar el **tamaño muestral** contribuye a reducir ambos errores.

# **Actividad 1 — Control de temperatura en un laboratorio**

Un laboratorio biotecnológico establece que la temperatura de sus salas de incubación debe mantenerse en 22 °C para asegurar condiciones óptimas de desarrollo celular. Con el fin de verificar este estándar, se tomaron 25 mediciones de temperatura, obteniendo una media de 22,5 °C y una desviación estándar de 0,8 °C. La dirección técnica solicita comprobar, al 5% de significancia, si la temperatura media realmente se mantiene en 22 °C o si existe una diferencia significativa respecto a este valor de referencia.

a. Formule las hipótesis

**Hipótesis**  

- $H_{0}$: $\mu = 22$
- $H_{1}$: $\mu \neq 22$

En lenguaje verbal
- $H_{0}$: La temperatura promedio real de las salas es igual a $22 °C$
- $H_{1}$: La temperatura promedio real de las salas es distinta a $22 °C$



b. Identifique qué distribución de probabilidad debe utilizar ($Z$ o $t$ de Student) justificando su elección. Luego, calcule el estadístico de prueba.

Se usa $t$ de Student porque la desviación poblacional $\sigma$ es desconocida y se trabaja con la desviación muestral $s = 0,8$ con $n = 25$.

```python
import numpy as np

x_bar = 22.5
mu0 = 22
s = 0.8
n = 25
alpha = 0.05

# Estadístico de prueba
t = (x_bar - mu0)/(s/np.sqrt(n))
print(f"Estadístico de prueba = {t:.3f}")
```

c) ¿Es una prueba unilateral o bilateral? ¿Cuántas regiones críticas tiene? justifique su respuesta.

Es una prueba bilateral, porque la alternativa es “distinto de 22°C”. Tiene 2 regiones críticas

d) Calcule el valor $p$

```python
import scipy.stats as st

# Cálculo valor p
p_valor = 2*(1-st.t.cdf(abs(t), df = n-1))

print(f"Valor p = {p_valor:.4f}")
```

e) Compare el valor $p$ con el nivel de significancia $(\alpha)$. ¿Se rechaza o no se rechaza $H_0$?

El valor de $t$ calculado es 3,125 con un p-valor = 0,0046, menor que el nivel de significancia $\alpha = 0,05$. Esto significa que existe evidencia estadística suficiente para rechazar la hipótesis nula.

f) Redacte una respuesta final en el contexto del problema, ¿se cumple con el estándar de temperatura?

Con un 5% de significancia, hay evidencia de que la temperatura media no es de 22°C. Por lo tanto, no se cumple con el estándar de 22°C.

```python
# OPCIONAL

import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt

# Parámetros
alpha = 0.05
df = 24
t_stat = 3.125
x = np.linspace(-4, 4, 500)

# Distribución t
y = st.t.pdf(x, df)

# Valores críticos
t_crit = st.t.ppf(1 - alpha/2, df)

# Gráfico
plt.figure(figsize=(8,4))
plt.plot(x, y, label="Distribución t (gl=24)")
plt.fill_between(x, y, 0, where=(x <= -t_crit) | (x >= t_crit), color="red", alpha=0.3, label="Región crítica")
plt.axvline(t_stat, color="blue", linestyle="--", label=f"t observado = {t_stat:.2f}")
plt.axvline(-t_crit, color="black", linestyle="--", label=f"± t crítico = {t_crit:.2f}")
plt.axvline(t_crit, color="black", linestyle="--")
plt.title("Prueba bilateral con regiones críticas")
plt.legend()
plt.show()
```

# **Actividad 2 — Tiempo de entrega en comercio electrónico**

Una empresa asegura que su tiempo medio de entrega es de a lo más 48 horas. Para auditar el cumplimiento, se tomó una muestra de 40 pedidos obteniendo un promedio de 50 horas. Por experiencia histórica se considera σ conocida igual a 6 horas (variabilidad del proceso). Se desea verificar con un nivel de significancia del 5% si, con esta evidencia, la empresa está incumpliendo la promesa (tiempos mayores a 48 h).

a. Formule las hipótesis

Hipótesis

- $H_{0}$: $\mu \leq 48$
- $H_{1}$: $\mu > 48$

En lenguaje verbal
- $H_{0}$: El tiempo promedio real de entrega es menor o igual a 48 horas.
- $H_{1}$: El tiempo promedio real de entrega es mayor a 48 horas.


b. Identifique qué distribución de probabilidad debe utilizar ($Z$ o $t$ de Student) justificando su elección. Luego, calcule el estadístico de prueba.

Se usa $Z$ porque la desviación estándar poblacional es conocida $\sigma=6$ horas.

```python
import numpy as np

mu0 = 48
x_bar = 50
sigma = 6
n = 40
alpha = 0.05

# Estadístico de prueba
z = (x_bar - mu0)/(sigma/np.sqrt(n))

print(f"Estadístico de Prueba = {z:.3f}")
```

c. ¿Es una prueba unilateral o bilateral? ¿Cuántas regiones críticas tiene? Justifique su respuesta.

Es unilateral derecha (se busca evidencia de que el promedio supera 48). Tiene 1 región crítica (cola derecha).

d. Calcule el valor $p$.

```python
# Cálculo valor p
p_valor = 1-st.norm.cdf(z)

print(f"Valor p = {p_valor:.4f}")
```

e. Compare el valor p con el nivel de significancia (α). ¿Se rechaza o no se rechaza $H_0$?

El estadístico es z = 2,108 y el p-valor = 0,0175. Al 5% de significancia, se rechaza $H_{0}$

f. Redacte una respuesta final en el contexto del problema, ¿la empresa está cumpliendo con el tiempo medio indicado?

Con 5% de significancia, hay evidencia de que el tiempo medio de entrega es mayor que 48 horas. Por lo tanto, la empresa no estaría cumpliendo su promesa de “a lo más 48 horas” con estos datos.

```python
# OPCIONAL

alpha = 0.05
df = 39
z_stat = 2.108
x = np.linspace(-4, 4, 500)

# Distribución normal
y = st.norm.pdf(x)

# Valor crítico
z_crit = st.norm.ppf(1 - alpha)

# Gráfico
plt.figure(figsize=(8,4))
plt.plot(x, y, label="Distribución normal estándar")
plt.fill_between(x, y, 0, where=(x >= z_crit), color="red", alpha=0.3, label="Región crítica")
plt.axvline(z_stat, color="blue", linestyle="--", label=f"z observado = {z_stat:.2f}")
plt.axvline(z_crit, color="black", linestyle="--", label=f"z crítico = {z_crit:.2f}")
plt.title("Prueba unilateral derecha con región crítica")
plt.legend()
plt.show()
```

# **Actividad 3 — Rendimiento académico**

En un curso universitario se sabe, por registros de años anteriores, que la primera prueba suele tener un promedio histórico de 55 puntos. Este año, después de implementar nuevas estrategias de enseñanza, se aplicó la misma prueba a 36 estudiantes, obteniendo una media muestral de 58,2 puntos con desviación estándar de 10 puntos. Se desea comprobar, con un nivel de significancia del 5%, si el rendimiento de los estudiantes ha mejorado significativamente respecto al promedio histórico.

a. Formule las hipótesis nula y alternativa

Hipótesis

- $H_{0}$: $\mu \leq 55$
- $H_{1}$: $\mu > 55$

En lenguaje verbal
- $H_{0}$: El rendimiento promedio real de los estudiantes es menor o igual a 55 puntos.
- $H_{1}$: El rendimiento promedio real de los estudiantes es mayor a 55 puntos.

b. Seleccione el modelo estadístico ($Z$ o $t$), justifique su elección y realice en Python el contraste de hipótesis.

```python
mu0 = 55
x_bar = 58.2
s = 10
n = 36
alpha = 0.05

# Estadístico de prueba
t = (x_bar - mu0)/(s/np.sqrt(n))

# Cálculo valor p
p_valor = 1-st.t.cdf(t, df=n-1)

# Resultados
print(f"t = {t:.3f}, p-valor = {p_valor:.4f}")
```

¿Existe evidencia al 5% de significancia para afirmar que la proporción real es mayor al 60%?

a. Formule las hipótesis nula y alternativa de la situación planteada.

Hipótesis

- $H_{0}: p \leq 0,60$
- $H_{1}: p > 0,60$

Lenguaje verbal
- $H_{0}:$ La proporción real de transporte público es menor o igual al 60%.
- $H_{1}:$ La proporción real de transporte público es mayor al 60%.

b. Determine la proporción muestral $\hat{p}$ de estudiantes que utilizan el transporte público.

```python
mu0_2, alpha_2 = 140, 0.05

# Estadístico de prueba
t2 = (xbar_c - mu0_2) / (s_c / np.sqrt(m_c))

# Cálculo valor p
p_valor2 = st.t.cdf(t2, df=m_c-1)

# Resultados
print(f"t = {t2:.3f}, p-valor = {p_valor2:.4f}")
```

# **Actividad 14 — Puntualidad de buses**

La empresa de transporte urbano se ha comprometido públicamente a que al menos 88 % de sus buses salgan puntuales para mejorar la experiencia de los pasajeros y cumplir con estándares de servicio. Con el fin de verificar si realmente se está cumpliendo este nivel de puntualidad, se realiza una auditoría sobre 220 salidas de buses durante la última semana. El informe indica que 206 salidas fueron puntuales.

Se desea comprobar, con un nivel de significancia $\alpha = 0.1$, si la puntualidad real de los buses es distinta al compromiso del 88 %. De ser así, la empresa podría respaldar su comunicación institucional con evidencia estadística. Luego de realizar el contraste de hipótesis correspondiente, ¿qué conclusión operativa debería comunicar la empresa?

*Desarrollo*

Hipótesis
- $H_{0}: p = 0.88$
- $H_{1}: p \neq 0.88$

Lenguaje verbal
- $H_{0}:$ La proporción real de buses puntuales es igual al 88%
- $H_{1}:$ La proporción real de buses puntuales es distinta del 88%


```python
import numpy as np
from scipy.stats import norm

# Datos
n = 220
x = 206
p_hat = x / n
p0 = 0.88
alpha = 0.10

# Estadístico de prueba
z = (p_hat - p0) / np.sqrt(p0 * (1 - p0) / n)

# Cálculo valor p
p_valor = 2 * (1 - norm.cdf(abs(z)))

# Resultados
print(f"p = {p_hat:.3f}")
print(f"z = {z:.2f}, p-valor = {p_valor:.4f}")
```

**Interpretación**

El p-valor ≈ 0.01 es menor que el nivel de significancia 0.10; por lo tanto, se rechaza $H_{0}$. Hay evidencia estadística de que la puntualidad de los buses es distinta al 88 % comprometido.
