## **Recordatorio Clave**

El modelo de *Regresión Lineal Simple* permite expresar la relación entre una variable dependiente $Y$ y una variable independiente $X$:

$$ Y_{i} = \beta_{0} + \beta_{1}X_{i} + \varepsilon_{i}$$

donde:

$Y_{i}$ : Variable dependiente o respuesta

$X_{i}$ : Variable independiente o predictora

$\beta_{0}$ : intercepto, valor esperado de $Y$ cuando $X = 0$

$\beta_{1}$ : pendiente, variación promedio en $Y$ por cada unidad de $X$

$\varepsilon_{i}$ : término de error (parte no explicada por el modelo)


### Supuestos del Modelo

Para interpretar un modelo de regresión lineal simple, es importante considerar que la relación entre las variables debe poder representarse razonablemente mediante una recta. Además, las predicciones deben realizarse con cautela, especialmente cuando se utilizan valores de Xalejados del rango observado en los datos. La calidad del modelo debe analizarse considerando tanto sus indicadores estadísticos como la interpretación contextual de sus resultados.

1. Relación lineal entre $X$ e $Y$
2. Errores con media cero y varianza constante (homocedasticidad)
3. Errores independientes y con distribución normal

### Contraste de hipótesis para la pendiente

Si el valor $p < α$, se rechaza $H_0$. En ese caso, existe evidencia estadística para afirmar que hay una relación lineal significativa entre $X$ e $Y$, dentro del contexto analizado.


### Indicadores importantes

| Estadístico | Significado |
|--------------|-------------|
| Valor estimado $(\widehat{Y})$ | Predicción de la variable dependiente obtenida a partir del modelo. |
| Intercepto  $(β_0)$ | Valor estimado de $Y$ cuando $X=0$. Debe interpretarse con cuidado, especialmente si $X=0$ está fuera del rango observado.|
| Pendiente $(\beta_{1})$ | Indica cuánto cambia, en promedio, $(Y)$ cuando $(X)$ aumenta en una unidad. |
| Valor $p$ de la pendiente | Permite evaluar si existe evidencia estadística de una relación lineal entre $(X)$ e $(Y)$. |
| Coeficiente de determinación $(R^2)$ | Indica qué proporción de la variabilidad de (Y) es explicada por el modelo lineal. |
| Residuo $(Y - \widehat{Y})$ | Diferencia entre el valor observado y el valor estimado por el modelo. Permite analizar los errores de estimación. |

# **Actividad 1 - Relación entre edad y gasto en alimentos**

El área de estudios de mercado de una empresa de consumo masivo busca analizar si existe una relación lineal entre la edad de las personas y su gasto mensual en alimentos.
Para ello, se seleccionó una muestra de 10 clientes con distintos rangos de edad, registrando su gasto promedio mensual en dólares. El objetivo es determinar si existe una relación lineal significativa entre ambas variables y, en caso afirmativo, estimar el modelo de regresión que describa dicha relación.


\begin{array}{|c|c|}
\hline
\textbf{Edad (años)} & \textbf{Gasto (USD)} \\ \hline
20 & 150 \\ \hline
25 & 180 \\ \hline
30 & 200 \\ \hline
35 & 220 \\ \hline
40 & 230 \\ \hline
45 & 250 \\ \hline
50 & 270 \\ \hline
55 & 290 \\ \hline
60 & 310 \\ \hline
65 & 330 \\ \hline
\end{array}


Con la información responda:

1. Identifique las variables:
* ¿Cuál es la variable dependiente?
* ¿Cuál es la variable independiente?


- Variable dependiente (y): Gasto mensual en alimentos (USD)
- Variable independiente (x): Edad en años

2. Escriba la ecuación del modelo estimado.

```python
import numpy as np
from scipy import stats

edad  = [20, 25, 30, 35, 40, 45, 50, 55, 60, 65]
gasto = [150, 180, 200, 220, 230, 250, 270, 290, 310, 330]

slope, intercept, r, p, se = stats.linregress(edad, gasto)

print(f"Modelo estimado: Ŷ = {intercept:.4f} + {slope:.4f} * X")
print(f"R²: {r**2:.4f}")
print(f"Valor p: {p:.6f}")
```

3. Grafique los datos y dibuje en el mismo gráfico la recta de regresión obtenida.

```python
import matplotlib.pyplot as plt
import numpy as np

edad  = [20, 25, 30, 35, 40, 45, 50, 55, 60, 65]
gasto = [150, 180, 200, 220, 230, 250, 270, 290, 310, 330]

x = np.array(edad)
y_hat = intercept + slope * x

plt.scatter(edad, gasto, label="Datos observados")
plt.plot(x, y_hat, color="red", label=f"Ŷ = {intercept:.2f} + {slope:.2f}X")
plt.xlabel("Edad (años)")
plt.ylabel("Gasto mensual (USD)")
plt.title("Relación entre edad y gasto en alimentos")
plt.legend()
plt.grid(True)
plt.show()
```

4. Interprete la pendiente y el intercepto en el contexto del problema.

Pendiente ($\hat{\beta}_1 = 3.8182$):
Por cada año adicional de edad, el gasto mensual en alimentos aumenta en promedio $3.82 USD.

Intercepto ($\hat{\beta}_0 = 80.7273$):
Cuando la edad es 0 años, el modelo estima un gasto de $80.73 USD. Sin embargo, esta interpretación no tiene sentido en el contexto del problema, ya que no existen clientes de 0 años en el estudio. El intercepto es solo un parámetro de ajuste de la recta.

5. Analice el valor $p$ asociado a la pendiente.


El valor $p$ es prácticamente 0, por lo tanto es menor que cualquier nivel de significancia usual ($\alpha = 0.05$ o $\alpha = 0.01$). Esto indica que se rechaza la hipótesis nula $H_0: \beta_1 = 0$, es decir, existe evidencia estadística suficiente para concluir que la edad tiene un efecto lineal significativo sobre el gasto mensual en alimentos.

6. ¿Qué proporción de la variabilidad del gasto se explica por la edad ($R^2$)?

El modelo explica el 99.53% de la variabilidad del gasto mensual en alimentos a partir de la edad. Esto indica que el modelo de regresión lineal simple tiene un ajuste muy bueno para estos datos.

7. Interprete el coeficiente de determinación $R^2$

# **Actividad 3 — Predicción del ingreso mensual**

Una institución financiera está interesada en analizar si el nivel educativo (expresado en años de estudio) puede utilizarse para predecir el ingreso mensual promedio de sus clientes.

El objetivo es determinar si existe una relación lineal significativa entre ambas variables y, en caso afirmativo, utilizar el modelo para estimar ingresos esperados en función del nivel educativo alcanzado.

Los datos recopilados corresponden a una muestra de clientes con distintos niveles de educación formal. Para cada nivel de años de estudio, se registra el ingreso mensual promedio, medido en dólares.


\begin{array}{|c|c|}
\hline
\textbf{Años de estudio} & \textbf{Ingreso (USD)} \\ \hline

10 & 500 \\ \hline
12 & 650 \\ \hline

A partir de estos datos, responda las siguientes preguntas utilizando Python como herramienta de apoyo:

# **Actividad 4 — Análisis del valor p y nivel de significancia**

Con los resultados de la Actividad 3:

1. ¿Qué representa el valor p asociado a la pendiente en un modelo de regresión?  

El valor $p$ asociado a la pendiente representa la probabilidad de obtener un coeficiente tan extremo como el observado, suponiendo que la hipótesis nula ($H_0: \beta_1 = 0$) es verdadera, es decir, que no existe relación lineal entre las variables. Un valor $p$ muy pequeño indica que es muy poco probable que la pendiente sea cero, por lo que hay evidencia para afirmar que la variable independiente sí influye sobre la dependiente.

2. ¿Qué implica que el valor $p$ sea menor que 0,05?  

```python
alpha = 0.05
print(f"Valor p: {p:.6f}")
print(f"Nivel de significancia: {alpha}")
if p < alpha:
    print("Se rechaza H0: existe relación lineal significativa (p < 0.05)")
else:
    print("No se rechaza H0: no hay evidencia de relación lineal significativa (p >= 0.05)")
```

Como el valor $p \approx 0.000001 < 0.05$, se rechaza $H_0: \beta_1 = 0$. Existe evidencia estadística suficiente para concluir que el nivel educativo tiene un efecto lineal significativo sobre el ingreso mensual.

3. ¿Y si comparamos con un nivel de significancia de 0,10?

```python
alpha = 0.10
print(f"Valor p: {p:.6f}")
print(f"Nivel de significancia: {alpha}")
if p < alpha:
    print("Se rechaza H0: existe relación lineal significativa (p < 0.10)")
else:
    print("No se rechaza H0: no hay evidencia de relación lineal significativa (p >= 0.10)")
```

Como el valor $p \approx 0.000001 < 0.10$, se rechaza $H_0: \beta_1 = 0$. La conclusión es la misma que con $\alpha = 0.05$: el nivel educativo tiene un efecto lineal significativo sobre el ingreso mensual.

4. ¿La conclusión sobre la relación entre nivel educativo e ingreso es estable ante cambios en $\alpha$?


Sí, la conclusión es estable ante cambios en $\alpha$. Tanto con $\alpha = 0.05$ como con $\alpha = 0.10$, el valor $p \approx 0.000001$ es considerablemente menor que ambos niveles de significancia, por lo que en ambos casos se rechaza $H_0$. Esto indica que la relación lineal entre el nivel educativo y el ingreso mensual es robusta y no depende del nivel de significancia elegido.

# **Errores de estimación en un modelo de regresión**

Hasta ahora hemos trabajado en la construcción e interpretación de modelos de regresión lineal simple, analizando la relación entre una variable independiente y una variable dependiente. Sin embargo, no basta con obtener un modelo y sus coeficientes: también es importante observar los errores que comete el modelo al estimar la variable dependiente.

En regresión, estos errores se llaman residuos y corresponden a la diferencia entre los valores observados y los valores estimados por el modelo:

$$Residuo = Y - \widehat{Y}$$

El análisis visual de los residuos permite observar si los errores se distribuyen alrededor de cero o si presentan algún patrón evidente. Esta revisión es una primera aproximación al análisis de errores del modelo; existen otros criterios para evaluar con mayor profundidad la calidad del ajuste y el uso del modelo en contextos predictivos.

Luego de realizar un estudio de mercado y ajustar un modelo de regresión lineal simple, se obtuvo la siguiente ecuación:

El intercepto $\hat{\beta}_0 = 12.5$ indica que cuando el gasto en publicidad en redes sociales es 0 miles de USD, el modelo estima ventas de 12.5 miles de unidades. Esto podría interpretarse como las ventas base que la empresa obtiene sin ninguna inversión publicitaria en redes sociales.

2. Interprete el significado de la pendiente en el contexto del problema.

La pendiente $\hat{\beta}_1 = 2.3$ indica que por cada mil dólares adicionales invertidos en publicidad en redes sociales, las ventas aumentan en promedio 2.3 miles de unidades.

3. ¿Qué representa la predicción obtenida cuando $X = 5$ en el contexto de la empresa?

```python
x_nuevo = 5
y_estimado = 12.5 + 2.3 * x_nuevo
print(f"Ventas estimadas para X = {x_nuevo} miles de USD: {y_estimado} miles de unidades")
```

Cuando la empresa invierte 5 miles de USD en publicidad en redes sociales, el modelo estima ventas de 24 miles de unidades. Esto representa la combinación de las ventas base (12.5) más el efecto de la inversión publicitaria ($2.3 \times 5 = 11.5$).

# **Actividad 8 - Transmisión de información en redes informáticas**

En un departamento de informática, un grupo de investigación en telecomunicaciones desea analizar la relación entre la longitud de los ficheros enviados por una red y el tiempo que tardan en llegar a un servidor.

Para ello, se realizaron experimentos enviando ficheros de distintas longitudes, medidas en bytes, y registrando el tiempo de transmisión correspondiente, medido en milisegundos. Los resultados de los experimentos se encuentran en el set de datos adjunto. Descargue el set de datos **<a href="https://docs.google.com/spreadsheets/d/128qjclJ0AoNyv4pHJvF8HtsAVVg_qSum/export?format=xlsx">aquí</a>**.



1.	Obtenga el modelo de regresión lineal que permita estimar el tiempo de transmisión en función de la longitud del fichero.


```python
import pandas as pd
from scipy import stats

df = pd.read_excel("G6_A8_datos_transmision.xlsx")
longitud = df["longitud_bytes"]
tiempo   = df["tiempo_ms"]

slope, intercept, r, p, se = stats.linregress(longitud, tiempo)

print(f"Modelo estimado: Ŷ = {intercept:.4f} + {slope:.4f} * X")
print(f"R²: {r**2:.4f}")
print(f"Valor p: {p:.6f}")
```

$$\hat{Y} = 27.3935 + 0.3081 , X$$

Donde $\hat{Y}$ es el tiempo de transmisión estimado (ms) y $X$ es la longitud del fichero (bytes).

2. Grafique los datos y escriba en el mismo gráfico la recta de regresión obtenida.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(longitud)
y_hat = intercept + slope * x

plt.scatter(longitud, tiempo, alpha=0.3, label="Datos observados")
plt.plot(x, y_hat, color="red", label=f"Ŷ = {intercept:.2f} + {slope:.4f}X")
plt.xlabel("Longitud del fichero (bytes)")
plt.ylabel("Tiempo de transmisión (ms)")
plt.title("Relación entre longitud del fichero y tiempo de transmisión")
plt.legend()
plt.grid(True)
plt.show()
```

3. Determine e interprete el coeficiente de determinación.

El modelo explica el 65.79% de la variabilidad del tiempo de transmisión a partir de la longitud del fichero. Esto indica un ajuste moderado, lo que sugiere que la longitud del fichero es un predictor relevante pero no suficiente por sí solo para explicar completamente el tiempo de transmisión, existiendo otros factores que el modelo no considera.

4. ¿El modelo es significativo? Justifique.

Como el valor $p \approx 0.000000 < 0.05$, se rechaza $H_0: \beta_1 = 0$ Si, el modelo es significativo: existe evidencia estadística suficiente para concluir que la longitud del fichero tiene un efecto lineal significativo sobre el tiempo de transmisión.

5. ¿Cuál será el tiempo de transmisión para un fichero que tiene una longitud de 250 bytes?

```python
x_nuevo = 250
y_estimado = intercept + slope * x_nuevo
print(f"Tiempo estimado para un fichero de {x_nuevo} bytes: {y_estimado:.4f} ms")
```

Para un fichero de 250 bytes, el modelo estima un tiempo de transmisión de $27.3935 + 0.3081 \times 250 = 104.42$ ms.
